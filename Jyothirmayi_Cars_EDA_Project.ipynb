```python
df_clean = df.copy()
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
print(df_clean.shape)   # should be (6926, 13)
```

```python
df_clean = df_clean.drop(columns=['torque'])
```

```python
df_clean['mileage_unit'] = df_clean['mileage'].str.extract(r'(kmpl|km/kg)')[0]
df_clean['mileage']   = pd.to_numeric(df_clean['mileage'].str.extract(r'([\d.]+)')[0], errors='coerce')
df_clean['engine']    = pd.to_numeric(df_clean['engine'].str.extract(r'([\d.]+)')[0], errors='coerce')
df_clean['max_power'] = pd.to_numeric(df_clean['max_power'].str.extract(r'([\d.]+)')[0], errors='coerce')
df_clean.info()
```

```python
print(df_clean.isnull().sum())
cols = ['mileage', 'engine', 'max_power', 'seats']
print(df_clean[cols].isnull().all(axis=1).sum())   # rows missing all four
```

```python
df_clean = df_clean.dropna(subset=cols).reset_index(drop=True)
df_clean['seats'] = df_clean['seats'].astype(int)
print(df_clean.shape)
```

```python
df_clean['brand'] = df_clean['name'].str.split().str[0]
df_clean['car_age'] = df_clean['year'].max() - df_clean['year']
df_clean['brand'].value_counts()
```

```python
cols = ['mileage', 'engine', 'max_power', 'seats']
print(df_clean[cols].isnull().all(axis=1).sum())
```

```python
df_clean['brand'] = df_clean['name'].str.split().str[0]
print(df_clean['brand'].value_counts())
print(df_clean['year'].max())
print(df_clean[['selling_price','km_driven','year','engine','max_power']].describe())
```

```python
df_clean.shape
```

```python
# Fix brand names that were cut off by splitting on the first space
df_clean['brand'] = df_clean['brand'].replace({'Land': 'Land Rover', 'Ashok': 'Ashok Leyland'})
```

```python
print(df_clean.sort_values('km_driven', ascending=False)[['name','year','km_driven','selling_price']].head(10))
print(df_clean.sort_values('km_driven')[['name','year','km_driven','selling_price']].head(10))
```

```python
df_clean['age'] = 2020 - df_clean['year']   # 2020 is the newest year in the data
print(df_clean.groupby('mileage_unit')['mileage'].describe())
print(df_clean['fuel'].value_counts())
```

```python
df_clean.shape
```

```python
print(df_clean.columns.tolist())
```

```python
# Remove impossible km_driven values: >500,000 km (data entry errors) and 1 km on a 2011 car
before = df_clean.shape[0]
df_clean = df_clean[(df_clean['km_driven'] <= 500000) & (df_clean['km_driven'] > 1)]
print(before, '->', df_clean.shape[0])
```

```python
print((df_clean['mileage'] == 0).sum())
print(df_clean[df_clean['mileage'] == 0][['name','fuel','mileage']].head())
```

```python
print(df_clean.sort_values('selling_price', ascending=False)[['name','year','km_driven','selling_price']].head(5))
```

```python
# 'car_age' duplicates 'age' (both = 2020 - year), so remove it
df_clean = df_clean.drop(columns=['car_age'])
print(df_clean.columns.tolist())
```

```python
# Mileage of 0 kmpl is impossible (missing value coded as 0), only 15 rows, so drop them
before = df_clean.shape[0]
df_clean = df_clean[df_clean['mileage'] > 0]
print(before, '->', df_clean.shape[0])
```

```python
print(df_clean['transmission'].value_counts())
print(df_clean['seller_type'].value_counts())
print(df_clean['owner'].value_counts())
```

```python
# Numerical summary
num_cols = ['selling_price', 'km_driven', 'year', 'age', 'mileage', 'engine', 'max_power', 'seats']
print(df_clean[num_cols].describe().round(2))

# Categorical summary
for c in ['brand', 'fuel', 'seller_type', 'transmission', 'owner']:
    print(df_clean[c].value_counts(), '\n')

# Skewness shows how lopsided price is
print(df_clean['selling_price'].skew())
print(df_clean['selling_price'].mean(), df_clean['selling_price'].median())
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.hist(df_clean['selling_price'] / 100000, bins=50, edgecolor='black')
plt.title('Distribution of Used Car Selling Prices')
plt.xlabel('Selling price (lakh INR)')
plt.ylabel('Number of cars')
plt.savefig('chart1_price_hist.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
top_brands = df_clean['brand'].value_counts().head(10)
plt.figure(figsize=(8,5))
plt.barh(top_brands.index[::-1], top_brands.values[::-1], edgecolor='black')
plt.title('Top 10 Brands by Number of Listings')
plt.xlabel('Number of cars')
plt.ylabel('Brand')
plt.savefig('chart2_top_brands.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Only brands with at least 30 listings, so a single car cannot distort the average
brand_counts = df_clean['brand'].value_counts()
big_brands = brand_counts[brand_counts >= 30].index
avg_price = (df_clean[df_clean['brand'].isin(big_brands)]
             .groupby('brand')['selling_price'].mean().sort_values() / 100000)

plt.figure(figsize=(8,7))
plt.barh(avg_price.index, avg_price.values, edgecolor='black')
plt.title('Average Selling Price by Brand (brands with 30+ listings)')
plt.xlabel('Average selling price (lakh INR)')
plt.ylabel('Brand')
plt.savefig('chart3_avg_price_brand.png', dpi=150, bbox_inches='tight')
plt.show()
```
